# Network Incident Data Analysis\nThis notebook analyzes simulated firewall, VPN, and network incident logs using Python and Pandas.

```python
import pandas as pd\nimport matplotlib.pyplot as plt\nfrom pathlib import Path\n
```

```python
df = pd.read_csv('../data/network_incidents.csv')\ndf['date'] = pd.to_datetime(df['date'])\ndf.head()
```

```python
df.shape
```

## Incident Category Summary

```python
category_summary = df.groupby('issue_category').agg(incident_count=('incident_id','count'), avg_resolution_time_minutes=('resolution_time_minutes','mean')).round(2).reset_index().sort_values('incident_count', ascending=False)\ncategory_summary
```

## Severity Summary

```python
severity_summary = df.groupby('severity').agg(incident_count=('incident_id','count'), avg_resolution_time_minutes=('resolution_time_minutes','mean')).round(2).reset_index()\nseverity_summary
```

## Visualization

```python
category_summary.plot(x='issue_category', y='incident_count', kind='bar', legend=False, figsize=(10,5))\nplt.title('Incident Count by Issue Category')\nplt.xlabel('Issue Category')\nplt.ylabel('Incident Count')\nplt.xticks(rotation=45, ha='right')\nplt.tight_layout()\nplt.show()
```

## Conclusion\nThe analysis helps identify common network issue types, severity patterns, and resolution-time trends that can support network operations decision-making.
